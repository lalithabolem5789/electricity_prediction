```python
import pandas as pd
import numpy as np
from glob import glob
import os

import seaborn as sns
import matplotlib.pyplot as plt

from scipy import stats
```

```python
# Load weather dataset
weather_df = pd.read_csv("E:/Project/weather_daily_darksky.csv")  

# Check for missing values
print(weather_df.isnull().sum())
```

```python
weather_df = pd.DataFrame({'temperatureMaxTime':pd.date_range('2011-11-11 23:00:00',periods=882)})
weather_df['day'] = pd.to_datetime(weather_df['temperatureMaxTime']).dt.date
weather_df['time'] = pd.to_datetime(weather_df['temperatureMaxTime']).dt.time
weather_df
```

```python
weather_df_original = pd.read_csv("E:/Project/weather_daily_darksky.csv") 
```

```python
weather_prefinal = pd.concat([weather_df, weather_df_original], axis = 'columns')
weather_prefinal
```

```python
weather_prefinal = weather_prefinal.drop([
    'temperatureMaxTime',
    'temperatureMinTime', 'apparentTemperatureMinTime',
    'apparentTemperatureHighTime', 'sunsetTime',
    'sunriseTime', 'temperatureHighTime', 'uvIndexTime',
    'summary', 'temperatureLowTime',
    'apparentTemperatureMaxTime', 'apparentTemperatureLowTime'
], axis='columns')
```

```python
weather_prefinal
```

```python
#according to weather_drop_time.info(), here Datatype of icon and precipType columns are object and string data:
#now we require to use 'hot_encoding' by get_dummies function:
icon = pd.get_dummies(weather_prefinal.icon)
icon

preciptype = pd.get_dummies(weather_prefinal.precipType)
preciptype

weather_add_icon = pd.concat([weather_prefinal, icon ], axis = 'columns')
weather_add_icon

weather_add_precip = pd.concat([weather_add_icon, preciptype  ], axis = 'columns')
weather_add_precip

weather_final = weather_add_precip.drop(['precipType', 'icon'], axis = 'columns')
weather_final
```

```python
print(weather_final.info())  # Check data types and missing values
print(weather_final.head())  # View first few rows
```
