```python
import pandas as pd
import numpy as np
from glob import glob
import os

import seaborn as sns
import matplotlib.pyplot as plt

from scipy import stats
```

```python
energy_data = pd.read_csv("E:/Project/Submissions/Datasets/Energy_processed.csv")
weather_data = pd.read_csv("E:/Project/Submissions/Datasets/Weather_processed.csv")
```

```python
merged_data = pd.DataFrame.merge(energy_data, weather_data, left_on='day', right_on='day', how='inner')  # Use 'outer' if needed
#merge_dataset = pd.DataFrame.merge(daywise_data,weather_final, left_on='day', right_on='day', how='inner')
#merge_dataset
```

```python
merged_data
```

```python
# Ensure 'day' is in datetime format
merged_data['day'] = pd.to_datetime(merged_data['day'])

# Extract year, month, and day
merged_data['year'] = merged_data['day'].dt.year
merged_data['month'] = merged_data['day'].dt.month
merged_data['date'] = merged_data['day'].dt.day
```

```python
print(merged_data.head())
print(merged_data.columns)
```

```python
print("Daywise Data Date Range:", energy_data['day'].min(), "to", energy_data['day'].max())
print("Weather Data Date Range:", weather_data['day'].min(), "to", weather_data['day'].max())
```

```python
# Drop the 'Unnamed: 0' column
merged_data = merged_data.drop(columns=['Unnamed: 0'])

# Verify the columns
print(merged_data.columns)
```

```python
print(merged_data.isnull().sum())  # Check NaN count for each column
print("Total missing values:", merged_data.isnull().sum().sum())  # Total NaN values in the dataset
```

```python
print(merged_data.dtypes)
```

```python
# Convert boolean categorical features (0 or 1 encoding)
bool_columns = ["clear-day", "cloudy", "fog", "partly-cloudy-day", "partly-cloudy-night", 
                "wind", "rain", "snow"]
merged_data[bool_columns] = merged_data[bool_columns].astype(int)
```

```python
corr = merged_data.select_dtypes(include=['number']).corr()

correlation = corr['energy_sum'].sort_values(ascending=False)
print(correlation)
```

```python
merged_data.to_csv("E:/Project/Submissions/Datasets/Final_Data.csv", index=False)  # Saves without the index column
print(f"\nProcessed dataset saved as: Final_Data.csv")
```
